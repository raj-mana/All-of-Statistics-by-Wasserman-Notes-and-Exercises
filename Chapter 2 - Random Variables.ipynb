#2.4 Some important continuous distributions

$\operatorname{Gamma}(\alpha, \beta)$. To remember the density, $\Gamma(\alpha)$ in the denominator is in the name Gamma distribution. To remember $\beta^\alpha$, note the $e^{-x/\beta}$ is dimensionless, so $\beta$ has the same units as $x$ and $dx$. The density is $\frac{1}{\Gamma(\alpha) \beta^\alpha} x^{\alpha-1}e^{-x/\beta}$. When $\alpha$ is a positive integer, $\operatorname{Gamma}(\alpha, \beta)$ is a sum of $\alpha$ IID $\operatorname{Exp}(\beta)$'s.

$\chi^2_p$. If $Z_1,\dots,Z_p \sim N(0,1)$ are IID, then $\sum Z_i^2 \sim \chi^2_p$, i.e., $\chi^2_p$ is a sum of $p$ IID standard normals. Also $\chi^2_p \sim \text{Gamma}(p/2,2)$. Just remember that $\chi^2_p $ is a Gamma, but not a sum of IID exponentials. Its mean is $p$ but it cannot be $\text{Gamma}(p,1)$, so the next distribution to try is $\text{Gamma}(p/2,2)$.

$\operatorname{Beta}(\alpha,\beta)$. Remember that the density has factors $x$ and $(1-x)$ with exponents $\alpha-1$ and $\beta-1$. The normalizing factor is $\Gamma(\alpha + \beta) / \Gamma(\alpha) \Gamma(\beta)$.

$t_\nu$. Let $Z \sim N(0,1)$ and $U \sim \chi^2_\nu$ be independent. Then
$$ \frac{Z}{\sqrt{U/\nu}} \sim t_\nu $$
The denominator of the LHS can be viewed as an approximate standard deviation. Also
$$ f_{t_\nu}(x) = \frac{\Gamma(\frac{\nu + 1}2)}{\Gamma(\frac\nu 2)} \frac 1{(1 + \frac{x^2}\nu)^{(\nu+1)/2}}$$
We probably don't need to memorize this, but observe that $t_1$ is the Cauchy distribution.

#2.8 Conditional Distributions

The conditional probability density function is defined by
$$ f_{X|Y=y}(x) = f_{X|Y}(x|y) = \frac{f_{X,Y}(x,y)}{f_Y(y)} $$
assuming that $f_Y(y) > 0$. Then
$$ \P(X \in A | Y = y) = \int_A f_{X|Y=y}(x) \, dx $$

We have

$$ f_{X,Y}(x,y) = f_X(x) f_{Y|X=x}(y) $$

#2.10 Two important multivariate distributions

Multinomial. $p = (p_1,\dots,p_k)$, $p_i \ge 0$, $\sum p_i = 1$. $n$ draws (independent draws with replacement).

$x_i \ge 0$ integer, $\sum x_i = n$.

$$ f(x) = \binom{n}{x_1,\dots,x_k} p_1^{x_1}\cdots p_k^{x_k} $$,
where
$$ \binom{n}{x_1,\dots,x_k} = \frac{n!}{x_1!\cdots x_k!}$$
If $X = (X_1, \dots, X_k) \sim \text{Multinomial}(n,p)$, then $X_i \sim \text{Binomial}(n,p_i)$.

Multivariate normal.
$$ f(x;\mu,\sigma) = \frac{1}{(2\pi)^{k/2} |\Sigma|^{1/2}} \exp\left( - \frac 12 (x-\mu)^T \Sigma^{-1} (x-\mu) \right)$$

If $Z \sim N(0,I)$, then $\mu + \Sigma^{1/2} Z \sim N(\mu,\Sigma)$

If $X \sim N(\mu, \Sigma)$, then $\Sigma^{-1/2}(X - \mu) \sim N(0,I)$

If we partition $X = (X_a, X_b)$, $\mu = (\mu_a,\mu_b)$, $$\Sigma = \begin{bmatrix} \Sigma_{aa } & \Sigma_{ab} \\ \Sigma_{ba} & \Sigma_{bb}\end{bmatrix}$$

Marginal distribution of $X_a$ is $N(\mu_a, \Sigma_{aa})$

For a deterministic vector $a$ we have $ a^T X \sim N(a^T \mu, a^T \Sigma a) $

$ (X - \mu)^T \Sigma^{-1}(X-\mu) \sim \chi^2_{k}$. Compare to the exponent in the density of $N(\mu, \Sigma)$.

To find $ (X_b | X_a ) $, notice that $\cov(X_b, X_a) = \Sigma_{ba}$, and so $\cov(X_b - \Sigma_{ba} \Sigma_{aa}^{-1}X_a, X_a) = 0$.

Thus,
$$ \begin{align*} \E(X_b - \Sigma_{ba} \Sigma_{aa}^{-1}X_a | X_a) &= \E(X_b - \Sigma_{ba} \Sigma_{aa}^{-1}X_a ) &\text{(independence)}\\
&= \mu_b - \Sigma_{ba}\Sigma_{aa}^{-1} \mu_a \\
\E(X_b | X_a) = \mu_b - \Sigma_{ba}\Sigma_{aa}^{-1}(\mu_a - X_a)
&=  \mu_b + \Sigma_{ba}\Sigma_{aa}^{-1}(X_a - \mu_a)
\end{align*}
$$
and
$$
\begin{align*}
\var(X_b | X_a) &= \var(X_b - \Sigma_{ba} \Sigma_{aa}^{-1}X_a | X_a) \\
&= \var(X_b - \Sigma_{ba} \Sigma_{aa}^{-1}X_a) \\
&= \var(X_b) - 2 \cov(X_b, \Sigma_{ba} \Sigma_{aa}^{-1} X_a) + \var(\Sigma_{ba} \Sigma_{aa}^{-1} X_a) \\
&= \Sigma_{bb} - 2 \Sigma_{ba}\Sigma_{aa}^{-1}\Sigma_{ab} + \Sigma_{ba}\Sigma_{aa}^{-1}\Sigma_{aa} \Sigma_{aa}^{-1}\Sigma_{ab} \\
&= \Sigma_{bb} - \Sigma_{ba} \Sigma_{aa}^{-1}\Sigma_{ab}
\end{align*}
$$
Hence
$$(X_b | X_a) \sim N(\mu_b + \Sigma_{ba}\Sigma_{aa}^{-1}(X_a - \mu_a), \Sigma_{bb} - \Sigma_{ba}\Sigma_{aa}^{-1}\Sigma_{ab}) $$

(b) If $N \sim \text{Poisson}(\lambda)$, we toss a coin $N$ times, and we let $X$ denote the number of heads and $Y$ the number of tails.

Then
$$
\begin{align*}

&= \frac{e^{-\lambda}}{k!}\sum_{n=k}^\infty  \frac{(\lambda/2)^n}{(n-k)!} \\
&= \frac{e^{-\lambda}}{k!} \cdot (\lambda/2)^{k}\sum_{n=0}^\infty  \frac{(\lambda/2)^n}{n!}  \\
&= e^{-\lambda/2}\frac{(\lambda/2)^k}{k!} \\

&= \P(Y = \ell | N = k + \ell) \frac{\P(N = k+ \ell)}{\P(Y = \ell)} \\
&= \frac{\binom{k+\ell}\ell 2^{-(k+\ell)}}{e^{-\lambda/2} \frac{(\lambda/2)^\ell}{\ell!}} \cdot e^{-\lambda}\frac{\lambda^{k+\ell}}{(k+\ell)!} \\
&= e^{-\lambda/2}\frac{(\lambda/2)^k}{k!} \\
&= \P(X=k)
\end{align*}
$$
So $X,Y$ are independent.
